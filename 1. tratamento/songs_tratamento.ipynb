```python
import pandas as pd
```

```python
df_songs = pd.read_csv("../0. data/Músicas/spotify_songs.csv")
```

```python
#Excluindo colunas desnecessárias
colunas = ["track_id", "track_album_id", "playlist_id", "danceability", "key", "loudness", "mode", "instrumentalness", "valence", "tempo"]

df_songs = df_songs.drop(colunas, axis=1)
```

```python
renomear = {
    "track_name" : "nome_musica",
    "track_artist" : "artista",
    "track_popularity" : "popularidade_musica",
    "track_album_name" : "album_musica",
    "track_album_release_date" : "data_lançamento_album",
    "playlist_name" : "nome_playlist",
    "playlist_genre" : "genero_playlist",
    "playlist_subgenre" : "subgenero_playlist",
    "energy" : "energia",
    "speechiness" : "fala",
    "acousticness" : "acusticidade",
    "liveness" : "prob_aovivo",
    "duration_ms" : "duracao_ms"
}
```

```python
#traduzindo colunas
df_songs = df_songs.rename(columns=renomear)
```

```python

#convertendo coluna DURACAO_MS de milisegundo para minutos
df_songs["duracao_min"] = round(df_songs["duracao_ms"] / 60000, 2)

df_songs.drop(columns="duracao_ms", axis=1)
```
